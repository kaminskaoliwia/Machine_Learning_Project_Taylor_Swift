## Taylor Swift Spotify Dataset - data preprocessing
The data used comes from [Kaggle](https://www.kaggle.com/datasets/jarredpriester/taylor-swift-spotify-dataset). This is a 
passion project used to learn the workflow and the rules of a proper analysis. The notebooks are based on previously written code, that is now cleaned up and structurized.

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
data_dir = Path("data")
input_path = Path("..") / data_dir / "taylor_swift_spotify.csv"

songs = pd.read_csv(input_path)
print("Shape of the data:")
print(songs.shape)
print()
```

The songs count is 582 with 18 columns describing them. The final goal of this project is to create an ML model
that correctly predicts in which era (new Taylor vs old Taylor) the song was released, therefore we remove columns that do not serve an anlytical purpose,
such as realease_date or track_number. 

```python
songs_info = songs.drop(['Unnamed: 0', 'release_date', 'track_number', 'id', 'uri'], axis=1)
```

Due to deluxe albums having extra content voice memos are present in the dataset. They are early versions
of songs that already exist in the dataset and have 0 popularity, so they will not be taken into consideration.

```python
print(songs_info[songs_info['popularity'] == 0]['name'])
print()

songs_info = songs_info[songs_info['popularity'] != 0]
print("Shape after deleting the irrelevant columns and voice memos:")
print(songs_info.shape)
```

For convenience we are changing full names of the albums to their common abbreviations. We do not take into consideration "Taylor's Versions" (TV) as they will be used later for experimentation.

```python
album_full_names = ['THE TORTURED POETS DEPARTMENT: THE ANTHOLOGY', 
                    'Midnights (3am Edition)', 
                    'evermore (deluxe version)', 
                    'folklore (deluxe version)',
                    'Lover', 'reputation', 
                    '1989 (Deluxe)', 
                    'Red', 
                    'Speak Now', 
                    'Fearless (Platinum Edition)', 
                    'Taylor Swift (Deluxe Edition)',
                    
                    "1989 (Taylor's Version) [Deluxe]", 
                    "Speak Now (Taylor's Version)", 
                    "Red (Taylor's Version)", 
                    "Fearless (Taylor's Version)"]

songs_info = songs_info[songs_info['album'].isin(album_full_names)]
```

```python
album_short_names = ['ttpd', 'midnights', 'evermore', 'folklore', 'lover', 'rep', '1989', 'red', 'speak now', 'fearless', 'debut', 
                     '1989', 'speak now', 'red', 'fearless']

album_dict = dict(zip(album_full_names, album_short_names))
songs_info['album'] = songs_info['album'].replace(album_dict)
```

```python
songs_info.to_csv(f'../{data_dir}/taylor_songs_no_TV.csv', index=False)
```

### Column meaning

```python
display(songs_info.sort_values('popularity', ascending=False).head())
```

**acousticness** - from 0 to 1, where 1 means most acoustic

**danceability** - from 0 to 1, where 1 means most danceable, takes into consideration tempo, rhythm stability, beat strength and regularity

**energy** - from 0 to 1, where 1 means most energetic, takes into consideration dynamic range, perceived loudness, timbre and entropy

**loudness** - The overall loudness of a track in decibels (dB). Values range between -60 and 0 db.

**speechiness** - detects the presence of spoken words in a track. The more exclusively speech-like the recording (e.g. talk show, audio book, poetry), the closer to 1.0 the attribute value. Values above 0.66 describe tracks that are probably made entirely of spoken words. Values between 0.33 and 0.66 describe tracks that may contain both music and speech, either in sections or layered, including such cases as rap music. Values below 0.33 most likely represent music and other non-speech-like tracks.

**tempo** - The overall estimated tempo of a track in beats per minute (BPM).

**valence** - A measure from 0.0 to 1.0 describing the musical positiveness conveyed by a track. Tracks with high valence sound more positive (e.g. happy, cheerful, euphoric), while tracks with low valence sound more negative (e.g. sad, depressed, angry).
